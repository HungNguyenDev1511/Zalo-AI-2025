```python
# Cell 1: Cài đặt Thư viện và Imports

import os
import json
import cv2
import numpy as np
import torch
from PIL import Image
from tqdm import tqdm
from peft import LoraConfig, prepare_model_for_kbit_training, get_peft_model
from transformers import (
    AutoProcessor,
    BitsAndBytesConfig,
    Idefics3ForConditionalGeneration,
    TrainingArguments,
    Trainer
)

# Kích hoạt mount Drive (Nếu chạy trên Google Colab)
# from google.colab import drive
# drive.mount('/content/drive')
```

```python
# Cell 2: Cấu hình Đường dẫn và Định nghĩa Hàm Xử lý Video

# Vui lòng cập nhật các đường dẫn này cho phù hợp với cấu trúc của bạn
DRIVE_ROOT = "/content/traffic_buddy_train+public_test"
TRAIN_JSON_PATH = f"{DRIVE_ROOT}/train/train.json"
TRAIN_VID_ROOT = f"{DRIVE_ROOT}/train/videos"

# Cấu hình Frames và Image
NUM_FRAMES = 5
TARGET_HEIGHT = 224

def extract_frames(video_path, num_frames=NUM_FRAMES):
    """Trích xuất N khung hình được phân bố đều từ một video."""
    frames = []
    try:
        cap = cv2.VideoCapture(video_path)
        if not cap.isOpened(): return []
        total = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
        if total == 0: cap.release(); return []
        idxs = np.linspace(0, total - 1, num_frames + 2, dtype=int)[1:-1]
        for i in idxs:
            cap.set(cv2.CAP_PROP_POS_FRAMES, i)
            ok, fr = cap.read()
            if ok and fr is not None:
                fr = cv2.cvtColor(fr, cv2.COLOR_BGR2RGB)
                frames.append(Image.fromarray(fr))
        cap.release()
    except Exception as e:
        # print(f"Lỗi khi xử lý video {video_path}: {e}")
        return []
    return frames

def merge_frames(frames):
    """Hợp nhất các khung hình PIL thành một ảnh duy nhất theo chiều ngang."""
    if not frames: return None

    resized_frames = [f.resize((int(f.size[0] * TARGET_HEIGHT / f.size[1]), TARGET_HEIGHT))
                      if f.size[1] != TARGET_HEIGHT else f for f in frames]

    widths, heights = zip(*(f.size for f in resized_frames))
    total_width = sum(widths)
    max_height = max(heights)

    merged = Image.new("RGB", (total_width, max_height))
    x_offset = 0
    for f in resized_frames:
        merged.paste(f, (x_offset, 0))
        x_offset += f.size[0]
    return merged
```

```python
# Cell 3: Tạo Tập Dữ liệu Huấn luyện (train_ds)

from datasets import Dataset # Import Dataset class

def get_answer_text(item):
    """Chuyển đổi đáp án chữ cái (A, B) thành nội dung (Xe tải)."""
    ans_letter = item["answer"].strip()
    for choice in item["choices"]:
        if choice.startswith(f"{ans_letter}."):
            return choice.split(".", 1)[-1].strip()
    return ans_letter

def create_train_dataset(train_json_path, train_vid_root, num_frames):
    """Tải dữ liệu JSON, xử lý video, và định dạng lại cho Fine-tuning."""
    try:
        with open(train_json_path, "r", encoding="utf-8") as f:
            data_raw = json.load(f)["data"]
    except FileNotFoundError:
        print(f"❌ Lỗi: Không tìm thấy file JSON tại {train_json_path}.")
        return None

    processed_data = []
    print(f"🔍 Bắt đầu xử lý {len(data_raw)} video huấn luyện...")
    for item in tqdm(data_raw, desc="Processing Train Videos"):
        vid_id = item["id"]
        video_path = os.path.join(train_vid_root, os.path.basename(item["video_path"]))

        frames = extract_frames(video_path, num_frames)
        merged_image = merge_frames(frames)

        if merged_image is None:
            continue

        question = item["question"]
        choices_text = "\n".join(item["choices"])
        correct_answer_text = get_answer_text(item)

        user_prompt = f"Câu hỏi: {question}\nCác lựa chọn:\n{choices_text}"

        messages_list = [
            {"role": "user", "content": [
                    {"type": "text", "text": "Answer briefly and accurately. "},
                    {"type": "image"},
                    {"type": "text", "text": user_prompt}
                ]},
            {"role": "assistant", "content": [
                    {"type": "text", "text": correct_answer_text}
                ]}
        ]

        processed_data.append({
            "id": vid_id,
            "messages": messages_list,
            "image": merged_image,
        })

    print(f"✅ Đã xử lý thành công {len(processed_data)} mục cho Fine-tuning.")
    return Dataset.from_list(processed_data)

# Chạy để tạo Dataset
train_ds = create_train_dataset(TRAIN_JSON_PATH, TRAIN_VID_ROOT, NUM_FRAMES)
if train_ds is None:
    print("Quá trình Fine-tuning bị hủy do lỗi tải dữ liệu.")
    # Sử dụng exit() hoặc raise Exception nếu không muốn chạy tiếp
```

```python
# Cell 4: Tải và Cấu hình Mô hình (QLoRA)

model_id = "HuggingFaceTB/SmolVLM-Base"
processor = AutoProcessor.from_pretrained(model_id)

# Cấu hình Lượng tử hóa (4-bit)
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16
)

# Cấu hình LoRA
lora_config = LoraConfig(
    r=8, lora_alpha=8, lora_dropout=0.1,
    target_modules=['down_proj','o_proj','k_proj','q_proj','gate_proj','up_proj','v_proj'],
    use_dora=False, init_lora_weights="gaussian"
)

print(f"🔄 Tải mô hình cơ sở {model_id} và áp dụng QLoRA...")

model = Idefics3ForConditionalGeneration.from_pretrained(
    model_id, quantization_config=bnb_config, _attn_implementation="flash_attention_2", device_map="auto"
)
model.add_adapter(lora_config)
model.enable_adapters()
model = prepare_model_for_kbit_training(model)
model = get_peft_model(model, lora_config)

print("Số tham số có thể huấn luyện (chỉ LoRA adapters):", model.get_nb_trainable_parameters())
print("✅ Mô hình đã sẵn sàng cho huấn luyện.")
```

```python
# Cell 5: Định nghĩa Hàm Data Collator

def collate_fn(examples):
  """
  Chuyển đổi Messages và Image thành tokenized inputs.
  Quan trọng: Mask (ẩn) phần User Prompt, Image Token và Padding trong labels.
  """
  texts = []
  images = []
  image_token_id = processor.tokenizer.additional_special_tokens_ids[
              processor.tokenizer.additional_special_tokens.index("<image>")]
  assistant_token_id = processor.tokenizer.convert_tokens_to_ids("<|assistant|>")

  for example in examples:
      messages = example["messages"]
      image = example["image"]
      text = processor.apply_chat_template(messages, add_generation_prompt=False)
      texts.append(text.strip())
      images.append([image])

  batch = processor(text=texts, images=images, return_tensors="pt", padding=True)
  labels = batch["input_ids"].clone()
  labels[labels == processor.tokenizer.pad_token_id] = -100

  for i in range(labels.size(0)):
      assistant_token_pos = (batch["input_ids"][i] == assistant_token_id).nonzero(as_tuple=True)[0]
      if len(assistant_token_pos) > 0:
          labels[i, :assistant_token_pos[0]] = -100

  labels[labels == image_token_id] = -100
  batch["labels"] = labels.to(torch.long)

  return batch
```

```python
# ⚙️ Cell 6: TrainingArguments và Trainer (tối ưu hóa cho A100)

model_name_base = model_id.split("/")[-1]
output_dir = f"./{model_name_base}-traffic-buddy-ft-v3.1"

# Tắt use_cache để tránh xung đột checkpointing và tối ưu tốc độ
model.config.use_cache = False

training_args = TrainingArguments(
    num_train_epochs=1,
    per_device_train_batch_size=8,  # tăng batch do A100 có 80GB
    gradient_accumulation_steps=2,  # batch ảo = 16
    warmup_steps=50,
    learning_rate=2e-4,
    weight_decay=0.01,
    logging_steps=50,
    save_strategy="epoch",  # chỉ lưu mỗi epoch, giảm I/O
    save_total_limit=1,
    optim="paged_adamw_8bit",  # dùng 8bit optimizer
    bf16=True,
    output_dir=output_dir,
    report_to="tensorboard",
    remove_unused_columns=False,
    gradient_checkpointing=False,  # 🚀 tắt để tăng tốc
    dataloader_num_workers=4,  # multi-thread data loading
)

trainer = Trainer(
    model=model,
    args=training_args,
    data_collator=collate_fn,
    train_dataset=train_ds,
)

print(f"✅ Trainer tối ưu cho A100 — batch_size={training_args.per_device_train_batch_size}, "
      f"accum={training_args.gradient_accumulation_steps}, bf16={training_args.bf16}, "
      f"checkpointing={training_args.gradient_checkpointing}")
```

```python
# Cell 7: Bắt đầu Huấn luyện và Lưu Mô hình

print("\n🚀 BẮT ĐẦU QUÁ TRÌNH FINE-TUNING...")

# Khởi chạy huấn luyện
trainer.train()

# Lưu mô hình Fine-tuned
trainer.save_model(output_dir)
print(f"\n✅ HOÀN TẤT! Mô hình đã được lưu tại {output_dir}")
```

```python
# ==============================================================================
# BƯỚC 1: CÀI ĐẶT & CẤU HÌNH INFERENCE
# ==============================================================================
import os, json, csv, torch, difflib, cv2
import numpy as np
from PIL import Image
from tqdm import tqdm
from transformers import AutoProcessor, BitsAndBytesConfig, Idefics3ForConditionalGeneration
from peft import PeftModel # Cần thư viện này để load adapter

# CẬP NHẬT đường dẫn adapter nếu bạn đổi phiên bản
MODEL_NAME = "HuggingFaceTB/SmolVLM-Base"
ADAPTER_PATH = f"./{MODEL_NAME.split('/')[-1]}-traffic-buddy-ft-v3.1"

# Cấu hình test data
DRIVE_ROOT = "/content/traffic_buddy_train+public_test"
TEST_JSON = f"{DRIVE_ROOT}/public_test/public_test.json"
TEST_VID_ROOT = f"{DRIVE_ROOT}/public_test/videos"
OUTPUT_CSV = "submission.csv"

NUM_FRAMES = 5
DEVICE = "cuda" if torch.cuda.is_available() else "cpu"

# ==============================================================================
# BƯỚC 2: HÀM TIỆN ÍCH
# ==============================================================================

def extract_frames(video_path, num_frames=5):
    frames = []
    try:
        cap = cv2.VideoCapture(video_path)
        if not cap.isOpened(): return []
        total = int(cap.get(cv2.CAP_PROP_FRAME_COUNT))
        if total == 0: cap.release(); return []
        idxs = np.linspace(0, total - 1, num_frames + 2, dtype=int)[1:-1]
        for i in idxs:
            cap.set(cv2.CAP_PROP_POS_FRAMES, i)
            ok, fr = cap.read()
            if ok and fr is not None:
                fr = cv2.cvtColor(fr, cv2.COLOR_BGR2RGB)
                frames.append(Image.fromarray(fr))
        cap.release()
    except Exception: return []
    return frames

def merge_frames(frames):
    if not frames: return None
    TARGET_HEIGHT = 224
    frames = [f.resize((int(f.size[0] * TARGET_HEIGHT / f.size[1]), TARGET_HEIGHT))
              if f.size[1] != TARGET_HEIGHT else f for f in frames]
    widths, heights = zip(*(f.size for f in frames))
    total_width = sum(widths)
    max_height = max(heights)
    merged = Image.new("RGB", (total_width, max_height))
    x_offset = 0
    for f in frames:
        merged.paste(f, (x_offset, 0))
        x_offset += f.size[0]
    return merged

def map_text_to_choice(answer_text, choices):
    if not isinstance(answer_text, str) or not answer_text.strip():
        return "A"
    text_upper = answer_text.strip().upper()

    # 1️⃣ Nếu model trả đúng A/B/C/D thì giữ nguyên
    if text_upper in ["A", "B", "C", "D"]:
        return text_upper

    # 2️⃣ Nếu có A., (B), v.v.
    for ch in ["A", "B", "C", "D"]:
        for pattern in [f"{ch}.", f"({ch})", f" {ch} ", f" {ch})"]:
            if pattern in text_upper:
                return ch

    # 3️⃣ So khớp nội dung
    cleaned_choices = []
    for opt in choices:
        if "." in opt:
            parts = opt.split(".", 1)
            if len(parts) > 1: opt = parts[1].strip()
        cleaned_choices.append(opt)
    scores = [difflib.SequenceMatcher(None, answer_text.lower(), o.lower()).ratio() for o in cleaned_choices]
    if scores and max(scores) > 0.85:
        return ["A","B","C","D"][int(np.argmax(scores))]

    # 4️⃣ fallback
    return "A"

# ==============================================================================
# BƯỚC 3: LOAD MODEL & INFERENCE
# ==============================================================================

bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16
)

print(f"🔁 Tải mô hình cơ sở {MODEL_NAME}...")
model_base = Idefics3ForConditionalGeneration.from_pretrained(
    MODEL_NAME,
    quantization_config=bnb_config,
    _attn_implementation="flash_attention_2",
    device_map="auto"
)
processor = AutoProcessor.from_pretrained(MODEL_NAME)

print(f"🔄 Tải và hợp nhất LoRA Adapter từ: {ADAPTER_PATH}")
model = PeftModel.from_pretrained(model_base, ADAPTER_PATH)
model = model.merge_and_unload()
model.eval()
print("✅ Model đã sẵn sàng.")

# Load dữ liệu test
with open(TEST_JSON, "r", encoding="utf-8") as f:
    data = json.load(f)["data"]
print(f"✅ Đã tải {len(data)} mục test.")

rows = []
for idx, item in enumerate(tqdm(data, desc="🔍 Inference")):
    vid_id = item["id"]
    q_text = item["question"]
    choices = item["choices"]
    video_path = os.path.join(TEST_VID_ROOT, os.path.basename(item["video_path"]))

    frames = extract_frames(video_path, NUM_FRAMES)
    merged_img = merge_frames(frames)
    if merged_img is None:
        rows.append((vid_id, "A"))
        continue

    # ✅ Tạo choices_text ở đây (fix bug NameError)
    choices_text = "\n".join([f"{chr(65+i)}. {c}" for i, c in enumerate(choices)])

    # Prompt buộc model trả về 1 ký tự
    user_prompt = (
        f"Dựa vào hình ảnh, hãy trả lời câu hỏi sau chỉ bằng **một chữ cái** (A, B, C hoặc D). "
        f"Không giải thích, không thêm ký tự nào khác.\n\n"
        f"Câu hỏi: {q_text}\nCác lựa chọn:\n{choices_text}\n\n"
        "👉 Chỉ trả lời bằng 1 chữ cái duy nhất."
    )

    messages = [
        {"role": "user", "content": [
            {"type": "text", "text": "Answer briefly and accurately."},
            {"type": "image"},
            {"type": "text", "text": user_prompt}
        ]}
    ]

    text_prompt = processor.apply_chat_template(messages, tokenize=False, add_generation_prompt=True)
    decoded = ""
    try:
        inputs = processor(text=text_prompt, images=[merged_img], return_tensors="pt", padding=True).to(DEVICE)

        with torch.no_grad():
            output = model.generate(
                **inputs,
                max_new_tokens=5,
                do_sample=False,
                eos_token_id=processor.tokenizer.eos_token_id,
                pad_token_id=processor.tokenizer.pad_token_id
            )
        decoded = processor.batch_decode(output, skip_special_tokens=True)[0]
        decoded = decoded.split("<|assistant|>")[-1].strip()

        # ✅ In debug cho 5 mẫu đầu tiên
        if idx < 5:
            print(f"\n[DEBUG] ID={vid_id}")
            print(f"Raw output: {decoded}")
            print(f"Choices: {choices}")

    except Exception as e:
        print(f"Lỗi Inference {vid_id}: {e}")

    ans_letter = map_text_to_choice(decoded, choices)
    rows.append((vid_id, ans_letter))

# ==============================================================================
# BƯỚC 4: LƯU CSV KẾT QUẢ
# ==============================================================================
with open(OUTPUT_CSV, "w", newline="", encoding="utf-8") as f:
    writer = csv.writer(f)
    writer.writerow(["id", "answer"])
    writer.writerows(rows)

print(f"\n✅ Done! Đã tạo {OUTPUT_CSV} ({len(rows)} hàng)")
```
